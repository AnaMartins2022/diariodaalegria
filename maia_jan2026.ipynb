```python
metros = [
    {"pos": [41.25495, -8.66801], "nome": "Metro Lidador"},
    {"pos": [41.24623, -8.6618], "nome": "Metro Pedras Rubras"},
    {"pos": [41.23825, -8.65821], "nome": "Metro Verdes"},
    {"pos": [41.23752, -8.6652], "nome": "Metro Botica"},
    {"pos": [41.23708, -8.66944], "nome": "Metro Aeroporto"},
    {"pos": [41.23301, -8.65659], "nome": "Metro Crestins"},
    {"pos": [41.26891, -8.61538], "nome": "Metro ISMAI"},
    {"pos": [41.26275, -8.61698], "nome": "Metro Castelo da Maia"},
    {"pos": [41.25358, -8.6283], "nome": "Metro Mandim"},
    {"pos": [41.24397, -8.62855], "nome": "Metro Zona Industrial"},
    {"pos": [41.23463, -8.62393], "nome": "Metro Forum Maia"},
    {"pos": [41.22902, -8.62689], "nome": "Metro Parque Maia"}
]

overpass_url = "http://overpass-api.de/api/interpreter"
# Query expandida para incluir Townhall
overpass_query = """
[out:json];
(
  node["amenity"~"school|kindergarten|childcare|hospital|clinic|doctors|pharmacy|police|bank|fire_station|townhall"](41.22, -8.68, 41.28, -8.60);
  way["amenity"~"school|kindergarten|childcare|hospital|clinic|doctors|pharmacy|police|bank|fire_station|townhall"](41.22, -8.68, 41.28, -8.60);
);
out center;
"""
response = requests.get(overpass_url, params={'data': overpass_query})
osm_data = response.json()

escolas_no_mapa = []
saude_no_mapa = []
farmacias_no_mapa = []
policia_no_mapa = []
bancos_no_mapa = []
bombeiros_no_mapa = []
camara_no_mapa = []

excluir_escolas = ["condução", "centro de estudo", "estudos", "cães", "canina", "driving"]

for element in osm_data.get('elements', []):
    lat = element.get('lat') or element.get('center', {}).get('lat')
    lon = element.get('lon') or element.get('center', {}).get('lon')
    nome_original = element.get('tags', {}).get('name', 'Equipamento')
    amenity = element.get('tags', {}).get('amenity')
    ponto = Point(lon, lat)

    dentro = False
    for coords in isocronas.values():
        if shape({"type": "Polygon", "coordinates": coords}).contains(ponto):
            dentro = True
            break

    if dentro:
        if amenity in ["school", "kindergarten", "childcare"]:
            if not any(word in nome_original.lower() for word in excluir_escolas):
                escolas_no_mapa.append({"pos": [lat, lon], "nome": nome_original})
        elif amenity in ["hospital", "clinic", "doctors"]:
            saude_no_mapa.append({"pos": [lat, lon], "nome": nome_original})
        elif amenity == "pharmacy":
            farmacias_no_mapa.append({"pos": [lat, lon], "nome": nome_original})
        elif amenity == "police":
            policia_no_mapa.append({"pos": [lat, lon], "nome": nome_original})
        elif amenity == "bank":
            bancos_no_mapa.append({"pos": [lat, lon], "nome": nome_original})
        elif amenity == "fire_station":
            bombeiros_no_mapa.append({"pos": [lat, lon], "nome": nome_original})
        elif amenity == "townhall":
            camara_no_mapa.append({"pos": [lat, lon], "nome": nome_original})

print(f"✅ Validados: {len(escolas_no_mapa)} Escolas, {len(saude_no_mapa)} Saúde, {len(farmacias_no_mapa)} Farmácias, {len(policia_no_mapa)} Polícia, {len(bancos_no_mapa)} Bancos, {len(bombeiros_no_mapa)} Bombeiros e {len(camara_no_mapa)} Edifício Municipal.")
```
